```python
RANDOM_SEED = 0

torch.manual_seed(RANDOM_SEED)
tensor_a = torch.rand(7,7)

# torch.manual_seed(RANDOM_SEED)
tensor_b = torch.rand(1,7)

res_mul = torch.matmul(tensor_a, tensor_b.T)
res_mul, res_mul.shape
```

```python
torch.cuda.is_available()
```

```python
torch.manual_seed(1234)
tensor_A = torch.rand(2,3)
# torch.cuda.manual_seed(1234)
tensor_B = torch.rand(2,3)

tensor_A, tensor_B
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

tensor_A.to(device), tensor_B.to(device)
```

```python
res_gpu = torch.matmul(tensor_A,tensor_B.T)
res_gpu, res_gpu.shape
```

```python
res_gpu.max(), torch.max(res_gpu)
```

```python
res_gpu.min(), torch.min(res_gpu)
```

```python
res_gpu.argmin(), torch.argmin(res_gpu)
```

```python
res_gpu.argmax(), torch.argmax(res_gpu)
```
